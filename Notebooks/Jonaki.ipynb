```python
import os

def print_tree(startpath):
    for root, dirs, files in os.walk(startpath):
        level = root.replace(startpath, '').count(os.sep)
        indent = '│   ' * (level)
        print(f"{indent}├── {os.path.basename(root)}/")
        subindent = '│   ' * (level + 1)
        for f in files:
            print(f"{subindent}├── {f}")

# Path to your specific folder
path = '/content/drive/MyDrive/Major Project'
print_tree(path)
```

```python
import os
import matplotlib.pyplot as plt

def draw_tree(startpath, output_img_name="project_structure.png"):
    lines = []
    for root, dirs, files in os.walk(startpath):
        level = root.replace(startpath, '').count(os.sep)
        indent = '    ' * (level)
        lines.append(f"{indent}📁 {os.path.basename(root)}/")
        subindent = '    ' * (level + 1)
        for f in files:
            lines.append(f"{subindent}📄 {f}")

    # Calculate image size dynamically based on line count
    line_count = len(lines)
    fig_height = max(4, line_count * 0.3)

    fig, ax = plt.subplots(figsize=(10, fig_height))
    ax.axis('off')

    # Render text line by line
    text_content = "\n".join(lines)
    ax.text(0.01, 0.99, text_content, fontfamily='monospace', fontsize=11,
            verticalalignment='top', horizontalalignment='left')

    # Save image to your notebook workspace
    plt.savefig(output_img_name, bbox_inches='tight', dpi=300)
    plt.close()
    print(f"Success! Image saved as '{output_img_name}' in your file sidebar.")

path = '/content/drive/MyDrive/Major Project'
draw_tree(path)
```
